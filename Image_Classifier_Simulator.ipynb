```python
import numpy as np
import matplotlib.pyplot as plt

# Step 1: Initialize Input Image Data
image = np.array([
    [1, 2, 1, 0, 2, 1],
    [2, 3, 2, 1, 3, 2],
    [1, 2, 4, 2, 1, 0],
    [0, 1, 2, 3, 2, 1],
    [2, 3, 1, 2, 4, 2],
    [1, 0, 2, 1, 2, 3]
])

print("Input Image:")
print(image)

# Step 2: Define Convolution Filter
kernel = np.array([
    [1, 0, -1],
    [1, 0, -1],
    [1, 0, -1]
])

# Step 3: Apply Convolution
def convolution(image, kernel):
    h, w = image.shape
    kh, kw = kernel.shape

    output = np.zeros((h-kh+1, w-kw+1))

    for i in range(h-kh+1):
        for j in range(w-kw+1):
            region = image[i:i+kh, j:j+kw]
            output[i, j] = np.sum(region * kernel)

    return output

conv_output = convolution(image, kernel)

print("\nConvolution Output:")
print(conv_output)

# Step 4: Apply Max Pooling
def max_pooling(matrix, size=2):
    h, w = matrix.shape
    output = np.zeros((h//size, w//size))

    for i in range(0, h-size+1, size):
        for j in range(0, w-size+1, size):
            region = matrix[i:i+size, j:j+size]
            output[i//size, j//size] = np.max(region)

    return output

pool_output = max_pooling(conv_output)

print("\nPooling Output:")
print(pool_output)

# Step 5: Display Feature Maps
plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(image, cmap='gray')
plt.title("Input Image")
plt.colorbar()

plt.subplot(1, 3, 2)
plt.imshow(conv_output, cmap='viridis')
plt.title("Convolution Feature Map")
plt.colorbar()

plt.subplot(1, 3, 3)
plt.imshow(pool_output, cmap='plasma')
plt.title("Max Pooling Output")
plt.colorbar()

plt.tight_layout()
plt.show()

print("\nImage Classifier Simulator Completed Successfully!")
```
